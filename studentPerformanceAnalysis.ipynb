```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Store CSV File data in DataFrame using Pandas

```python
data = pd.read_csv("student_data.csv")

print(data)
```

```python
print(data.describe())
```

Data Cleaning using Pandas

```python
print(data.isnull().sum())
```

handle null boxes

```python
data[["Math","Science","English"]] = data[["Math","Science","English"]].fillna(0) 
#consider zero marks for null marks 
data = data.dropna() 
#remove row if name or gender is missing
```

Create New column for Total and Average Marks

```python
data["Total"] = data["Math"] + data["Science"] + data["English"]
data["Average"]=data["Total"]/3
print(data.head())
```

Visualization Using Malplolib

Average Marks per Subject

```python
data[["Math","Science","English"]].mean().plot(kind="bar")

plt.title("Average Marks per Subject")
plt.ylabel("Marks")
plt.xlabel("subjects")
plt.show()
```

StudyHours vs Total Marks

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x="StudyHours",
    y="Total",
    data=data,
    s=120,
    color="#ff9f43",
    edgecolor="black"
)

plt.title("Study Hours vs Total Marks", fontsize=16, fontweight="bold")
plt.xlabel("Study Hours", fontsize=12)
plt.ylabel("Total Marks", fontsize=12)

plt.grid(True, linestyle="--", alpha=0.6)

plt.show()
```

Correlation Heatmap

```python
plt.figure(figsize=(8,6))

corr = data.select_dtypes(include="number").corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="RdYlBu",
    linewidths=1,
    linecolor="black",
    fmt=".2f"
)

plt.title("Feature Correlation Heatmap", fontsize=16, fontweight="bold")

plt.show()
```
